# mutant and wildtype epithelial only

```python
work_dir = '/share/crsp/lab/dalawson/tzhuravl/CellPlex_Tumors/Scanpy_Seurat/scVI_epithelial/mutant_epi'

adata = pickle.load(open(os.path.join(work_dir, 'no_schwann_mut_wt_epi_no_LumHR_scvi.pkl'), 'rb'))
```

```python
adata.obs=sub.obs.copy()
```

```python
adata.obs['cell_type'] = (
    adata.obs["cell_type"]
    .map(lambda x: {"tumor": "tumor-like"}.get(x, x))
    .astype("category")
)
```

```python
sc.pl.umap(adata, color='cell_type', size=25,
          palette = {'Bmyo':'#27AAE1','LumSec':'#2ca02c','tumor-like':'#C11C82'})
```

```python
not_mutant_cnv=adata[~adata.obs['ploidy'].isin(['aneuploid'])]
not_mutant_cnv


not_mutant_cnv.obs['cnv_or_cell_type']=not_mutant_cnv.obs['cell_type']

adata.obs['cnv_or_cell_type']=not_mutant_cnv.obs['cnv_or_cell_type']

categories = np.array(
     ['Bmyo', 'LumSec', 'tumor-like', 'Aneuploid'])

adata.obs['cnv_or_cell_type'] = pd.Categorical(
   adata.obs['cnv_or_cell_type'], categories=categories, ordered=True)

adata.obs['cnv_or_cell_type'].fillna("Aneuploid", inplace=True,)
```

```python
sc.pl.umap(adata, color='cnv_or_cell_type',size=25,
           palette=['#27AAE1','#2ca02c','#C11C82','black'])
```

```python
highlight = adata[adata.obs['orig_ident'].isin(['basal_wt'])]
adata.obs['highlight'] = highlight.obs['cnv_or_cell_type']

sc.pl.umap(adata, color='highlight', size=25,
          palette=['#27AAE1','#C11C82','black'])
```

```python
highlight = adata[adata.obs['orig_ident'].isin(['basal_mut'])]
adata.obs['highlight'] = highlight.obs['cnv_or_cell_type']

sc.pl.umap(adata, color='highlight', size=25,
          palette=['#27AAE1','#2ca02c','#C11C82','black'])
```

```python
highlight = adata[adata.obs['orig_ident'].isin(['luminal_wt'])]
adata.obs['highlight'] = highlight.obs['cnv_or_cell_type']

sc.pl.umap(adata, color='highlight', size=25,
          palette=['#2ca02c','#C11C82','black'])
```

```python
highlight = adata[adata.obs['orig_ident'].isin(['luminal_mut'])]
adata.obs['highlight'] = highlight.obs['cnv_or_cell_type']

sc.pl.umap(adata, color='highlight', size=25,
          palette=['#27AAE1','#2ca02c','#C11C82','black'])
```

```python
tmp = pd.crosstab(adata.obs['cnv_or_cell_type'],adata.obs['orig_ident'], 
                  normalize='columns').reindex(['Bmyo',
                                        'LumSec',
                                        'tumor-like',
                                                'Aneuploid'])[["basal_wt","basal_mut",
                                                               "luminal_wt","luminal_mut"]].T.plot(kind='bar',
                                                                                             figsize=(2,4), 
                                                                                             stacked=True,
                                                                                             color=['#27AAE1',
                                                                                                    '#2ca02c',
                                                                                                    '#C11C82',
                                                                                                    'black'])
tmp.legend(title='', bbox_to_anchor=(2, 1),loc='upper right')
plt.grid(None)
pl.xticks(rotation=45)
pl.savefig('fig4_cnv_or_cell_type_stacked.pdf', bbox_inches='tight')
```

```python
weights=pd.read_csv("/share/crsp/lab/dalawson/tzhuravl/Transplants/transplant_tumors_2batches.csv")
```

```python
weights=weights[weights['Transplant_cells'].isin(['Luminal','Basal'])]
```

```python

fig, ax = plt.subplots(figsize=(1.5, 4))

ax = sns.boxplot(x='Transplant_cells', y='Tumor_weight_grams', data=weights, 
                 showfliers=False,palette=['#2ca02c','#27AAE1']) # Hide default outliers

# 3. Overlay the swarm plot (or stripplot) to show colored points
# The 'hue' argument automatically colors points by category
sns.swarmplot(x='Transplant_cells', y='Tumor_weight_grams', data=weights, hue='Transplant_cells', ax=ax, 
              palette=['black','black'], marker='o', s=5)
        
plt.savefig('fig4_weights_barplots.pdf', bbox_inches='tight')
plt.show()
```
