```python
adatas = [adata_1, adata_2, adata_3, adata_4, adata_5, adata_6, adata_7]
adata = ad.concat(adatas,  join="outer")
adata
```

```python
sc.pl.highest_expr_genes(adata, n_top=20)
```

```python
sc.pp.filter_cells(adata, min_genes=500)
sc.pp.filter_genes(adata, min_cells=1)
adata
```

```python
adata.var["mt"] = adata.var_names.str.startswith("mt-")
adata.var["ribo"] = adata.var_names.str.startswith(("Rps", "Rpl"))
adata.var["hb"] = adata.var_names.str.startswith("Hbb")

sc.pp.calculate_qc_metrics(
    adata, qc_vars=["mt", "ribo", "hb"], inplace=True, log1p=True
)
```

```python
sc.pl.violin(
    adata,
    ["n_genes_by_counts", "total_counts", "pct_counts_mt"],
    jitter=0.4,
    multi_panel=True,
)
```

```python
sc.pl.violin(
    adata,
    ["pct_counts_mt", "pct_counts_ribo", "pct_counts_hb"],
    jitter=0.4,
    multi_panel=True,
)
```

```python
sc.pl.scatter(adata, x="total_counts", y="pct_counts_mt")
sc.pl.scatter(adata, x="total_counts", y="n_genes_by_counts")
```

```python
adata = adata[adata.obs.n_genes_by_counts < 6500, :]
adata = adata[adata.obs.total_counts < 40000, :]
adata = adata[adata.obs.pct_counts_mt < 6, :]
adata = adata[adata.obs.pct_counts_ribo < 40, :]
adata = adata[adata.obs.pct_counts_hb < 2, :].copy()
```

```python
sc.pl.violin(
    adata,
    ["n_genes_by_counts", "total_counts", "pct_counts_mt"],
    jitter=0.4,
    multi_panel=True,
)
```

```python
adata = adata[:, adata.var.highly_variable]
sc.pp.regress_out(adata, ['total_counts'])
sc.pp.scale(adata, max_value=10)
```

```python
sc.tl.pca(adata, svd_solver="arpack")
```

```python
sc.pl.pca_variance_ratio(adata, log=True)
```

```python
sc.pp.neighbors(adata, n_neighbors=10, n_pcs=40)
```

```python
sc.tl.leiden(
    adata,
    resolution=0.1,
    random_state=0,
    n_iterations=2,
    directed=False,
)
```

```python
sc.tl.paga(adata)
sc.pl.paga(adata)
sc.tl.umap(adata, init_pos='paga')
```
