```python
sc.pl.highest_expr_genes(adata, n_top=20)
```

```python
adata.var["mt"] = adata.var_names.str.startswith("mt-")
adata.var["ribo"] = adata.var_names.str.startswith(("Rps", "Rpl"))
adata.var["hb"] = adata.var_names.str.startswith("Hbb")

sc.pp.calculate_qc_metrics(
    adata, qc_vars=["mt", "ribo", "hb"], inplace=True, log1p=True
)
```

```python
adata = sc.read_h5ad('cellplex9.h5ad')
```

```python
sc.pp.normalize_total(adata, target_sum=1e4)
sc.pp.log1p(adata)
```

```python
adata.raw = adata
```

```python
adata = adata[:, adata.var.highly_variable]
sc.pp.regress_out(adata, ['total_counts'])
sc.pp.scale(adata, max_value=10)
```

```python
sc.tl.pca(adata, svd_solver="arpack")
```

```python
sc.pl.pca_variance_ratio(adata, log=True)
```

```python
sc.pp.neighbors(adata, n_neighbors=10, n_pcs=40)
```

```python
sc.tl.leiden(
    adata,
    resolution=0.1,
    random_state=0,
    n_iterations=2,
    directed=False,
)
```

```python
sc.tl.paga(adata)
sc.pl.paga(adata)
sc.tl.umap(adata, init_pos='paga')
```

```python
sc.pl.umap(adata, color=["Krt8", "Krt14", "Malat1", "Top2a"])
```

```python
sc.pl.umap(adata, color=["leiden", "Rgs5", "Ptprc", "Pecam1"])
```

```python
sc.pl.umap(
    adata,
    color=["leiden", "log1p_total_counts", "pct_counts_mt", "log1p_n_genes_by_counts"],
    wspace=0.5,
    ncols=2,
)
```

```python
for res in [0.1, 0.2, 0.5]:
    sc.tl.leiden(
        adata, key_added=f"leiden_res_{res:4.2f}", resolution=res
    )
```
