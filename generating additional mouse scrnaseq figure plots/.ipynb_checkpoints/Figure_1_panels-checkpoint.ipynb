# percentage of aneuploidy

```python
#wt
```

```python
wt = epi[epi.obs['condition'].isin(['normal'])]
df=wt.obs.copy()
df['Match_aneu'] = (df['cnv'] == "aneuploid").astype(int)
category_match = df.groupby('cell_type_final')['Match_aneu'].mean() * 100
rep_match = df.groupby(['cell_type_final','orig_ident'])['Match_aneu'].mean() * 100
rep_match=pd.DataFrame(rep_match)
#print(category_match)
```

```python
colors=['#27AAE1','#2ca02c','#C11C82']

plt.figure(figsize=(2, 6))
sns.barplot(x='cell_type_final', y='Match_aneu', data=rep_match, palette=colors)#, order=category_order)
plt.xlabel('Cell Type')
plt.ylabel('Percentage Aneuploid')
plt.title('Wildtype')
plt.tight_layout()
plt.xticks(rotation=90, ha='center')
plt.ylim(0, 100)
plt.savefig('percent_aneuploid_WT.pdf', format='pdf')
plt.show()
```
